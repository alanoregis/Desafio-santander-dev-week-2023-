```python
import requests
import json

sdw2023_api_url = 'http://localhost:8000'  


user_ids = [4, 5 , 6]  


def get_user(user_id):

    try:
        url = f'{sdw2023_api_url}/users/{user_id}'
        print(f"🔍 Buscando: {url}")
        
        response = requests.get(url, timeout=10)
        
        if response.status_code == 200:
            print(f"✅ Usuário {user_id} encontrado!")
            return response.json()
        elif response.status_code == 404:
            print(f"⚠️ Usuário {user_id} não existe (404)")
            return None
        else:
            print(f"⚠️ Status inesperado {response.status_code} para usuário {user_id}")
            return None
            
    except requests.exceptions.ConnectionError:
        print(f"❌ ERRO: Não foi possível conectar à API em {sdw2023_api_url}")
        print("   Verifique se a API está rodando:")
        print("   1. Execute: docker-compose up")
        print("   2. Ou teste manual: curl http://localhost:8000/users/1")
        return None
    except Exception as e:
        print(f"⚠️ Erro ao buscar usuário {user_id}: {type(e).__name__}")
        return None


print("=" * 60)
print("TESTANDO CONEXÃO COM A API SANTANDER DEV WEEK")
print("=" * 60)

print(f"🌐 API URL: {sdw2023_api_url}")
print(f"📋 IDs para buscar: {user_ids}")


test_id = 1
print(f"\n🧪 Teste rápido com ID {test_id}...")
try:
    test_response = requests.get(f"{sdw2023_api_url}/users/{test_id}", timeout=5)
    if test_response.status_code == 200:
        test_data = test_response.json()
        print(f"✅ API RESPONDE! Usuário teste: {test_data.get('name')}")
    else:
        print(f"⚠️ API respondeu com status: {test_response.status_code}")
except:
    print(f"❌ API NÃO RESPONDE em {sdw2023_api_url}")


print("\n" + "=" * 60)
print("BUSCANDO TODOS OS USUÁRIOS")
print("=" * 60)

users = [user for id in user_ids if (user := get_user(id)) is not None]

print("\n" + "=" * 60)
print("RESULTADOS FINAIS")
print("=" * 60)

if users:
    print(f"🎉 SUCESSO! {len(users)} de {len(user_ids)} usuários encontrados")
    
    
    print("\n📄 DADOS COMPLETOS (JSON):")
    print(json.dumps(users, indent=2, ensure_ascii=False))
    
  
    print("\n👥 RESUMO DOS USUÁRIOS:")
    for i, user in enumerate(users, 1):
        print(f"\n{i}. {user.get('name', 'Sem nome')} (ID: {user.get('id')})")
        
      
        account = user.get('account', {})
        if account:
            print(f"   📊 Conta: {account.get('number', 'N/A')}")
            print(f"   🏦 Agência: {account.get('agency', 'N/A')}")
            print(f"   💰 Saldo: R$ {account.get('balance', 0):.2f}")
            print(f"   🎯 Limite: R$ {account.get('limit', 0):.2f}")
        
    
        card = user.get('card', {})
        if card:
            print(f"   💳 Cartão: {card.get('number', 'N/A')}")
        
     
        news = user.get('news', [])
        if news:
            print(f"   📰 {len(news)} notícias personalizadas")
            
else:
    print("❌ NENHUM USUÁRIO ENCONTRADO")
    print("\n🔧 SOLUÇÃO:")
    print(f"   1. Verifique se a API está rodando em {sdw2023_api_url}")
    print(f"   2. Teste manual: curl {sdw2023_api_url}/users/1")
    print(f"   3. Verifique os IDs: {user_ids}")
    print("   4. Execute: docker-compose up (se usar Docker)")
```

```python
from openai import OpenAI  
import requests
import json


openai_api_key = "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" 

client = OpenAI(api_key=openai_api_key)


sdw2023_api_url = 'http://localhost:8000'


def generate_ai_news(user):
    try:
        completion = client.chat.completions.create(
            model="gpt-3.5-turbo",  
            messages=[
                {
                    "role": "system",
                    "content": "Você é um especialista em marketing bancário."
                },
                {
                    "role": "user",
                    "content": f"Crie uma mensagem para {user['name']} sobre a importância dos investimentos (máximo de 100 caracteres)"
                }
            ],
            max_tokens=100,  
            temperature=0.7  
        )
        
        return completion.choices[0].message.content.strip('"')
    
    except Exception as e:
        print(f"❌ Erro na OpenAI: {e}")
        return f"{user['name']}, invista no seu futuro! Pequenos passos geram grandes resultados."

if __name__ == "__main__":
    print("🔍 Buscando usuários da API Santander...")
    
    user_ids = [4, 5, 6]
    users = []
    
    for user_id in user_ids:
        try:
            response = requests.get(f'{sdw2023_api_url}/users/{user_id}')
            if response.status_code == 200:
                user = response.json()
                users.append(user)
                print(f"✅ {user['name']} carregado")
        except:
            print(f"⚠️ Erro ao carregar usuário {user_id}")
    
    if not users:
        print("❌ Nenhum usuário encontrado. Usando dados de exemplo...")
        users = [
            {"id": 1, "name": "João Silva", "news": []},
            {"id": 2, "name": "Maria Santos", "news": []}
        ]
    
    print("\n🤖 Gerando mensagens personalizadas com IA...")
    print("=" * 50)
    
    for user in users:
        print(f"\n👤 Processando: {user['name']}")
        
        news = generate_ai_news(user)
        print(f"📝 Mensagem: {news}")
        
        if 'news' not in user:
            user['news'] = []
        
        user['news'].append({
            "icon": "https://digitalinnovationone.github.io/santander-dev-week-2023-api/icons/credit.svg",
            "description": news
        })
        
        try:
            response = requests.put(f'{sdw2023_api_url}/users/{user["id"]}', json=user)
            if response.status_code == 200:
                print(f"💾 {user['name']} atualizado na API")
        except:
            print(f"⚠️ Não foi possível atualizar {user['name']}")
    
    print("\n" + "=" * 50)
    print("🎉 Processamento concluído!")
```

```python
def update_user(user):
  response = requests.put(f"{sdw2023_api_url}/users/{user['id']}", json=user)
  return True if response.status_code == 200 else False

for user in users:
  success = update_user(user)
  print(f"User {user['name']} updated? {success}!")
```
